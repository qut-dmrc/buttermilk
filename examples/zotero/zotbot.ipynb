```python
from rich import print

from buttermilk.utils.nb import *

job = "zotbot"
objs = nb_init(job=job, overrides=["llms=full", "+flows=[zot]"])
print(objs.flows)
```

```python
from buttermilk._core.config import AgentVariants
from buttermilk._core.contract import AgentInput
from buttermilk.agents.rag.rag_zotero import RagZotero

variants = AgentVariants(**objs.flows.zot.agents.rag_zotero).get_configs()
zotbots = []
for agent_cls, variant_config in variants:
    zotbots.append(RagZotero(**variant_config.model_dump()))
print(", ".join([z.agent_name for z in zotbots]))
```

```python
q = "What is digital constitutionalism?"
answer = await zotbots[2].invoke(AgentInput(inputs={"prompt": q}))

print(q)
print("Answer:")
print(answer)
```
